```python
##top 5 frequency words for the lab today 

#Hint: reduce by key only works with keys 


!pip install pyspark

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("RDD_Practice") \
    .getOrCreate()

sc = spark.sparkContext
```

```python
import string

text = sc.textFile("/Users/andreams/Downloads/sinners.txt")

stop_words = sc.textFile("/Users/andreams/Downloads/stop_words.txt")

stop_words_list = stop_words.collect()
```

```python
text_lowercase = text.map(lambda x: x.lower())

split_text = text_lowercase.flatMap(lambda x:x.split(" "))

no_punctuation_text = split_text.map(lambda x: x.strip(string.punctuation)) 

removed_stop_words = no_punctuation_text.filter(lambda x: x not in stop_words_list)

removed_stop_words.collect()
```

```python
word_count = removed_stop_words.map(lambda w:(w,1))

word_count.collect()
```

```python
word_count_reducebykey = word_count.reduceByKey(lambda x,y: (x+y))

word_count_reducebykey.collect()
```

```python
top_five_words = word_count_reducebykey.sortBy(lambda x: x[1], ascending=False)

top_five_words.collect()
```
